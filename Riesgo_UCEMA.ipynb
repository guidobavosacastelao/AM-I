# Teoría del Riesgo: Probabilidad de Ruina y Supervivencia

## 1. El proceso de reservas de una compañía aseguradora

En la Teoría del Riesgo Actuarial, uno de los principales objetivos es analizar la capacidad de una compañía aseguradora para afrontar sus obligaciones a lo largo del tiempo.

Para ello, se define el **proceso de reservas o superávit**, denotado por $U(t)$, que representa el capital disponible de la aseguradora en el instante $t$.

En el modelo clásico de riesgo de Cramér-Lundberg, este proceso se expresa como:

$$
U(t)=u+ct-\sum_{i=1}^{N(t)}X_i
$$

Donde:

- $u$: capital inicial de la aseguradora.
- $c$: ingreso por primas por unidad de tiempo.
- $N(t)$: número de siniestros ocurridos hasta el instante $t$.
- $X_i$: importe correspondiente al siniestro $i$.
- $U(t)$: capital disponible de la aseguradora en el instante $t$.

Por lo tanto, el capital aumenta con el cobro de primas y disminuye con el pago de siniestros.

El problema fundamental consiste en estudiar si, en algún momento, los pagos acumulados de siniestros pueden superar el capital inicial más las primas recibidas.

---

## 2. Tiempo de ruina

Se define el **tiempo de ruina**, denotado por $\tau$, como:

$$
\boxed{\tau=\inf\{t\geq 0:U(t)<0\}}
$$

Esta expresión representa el primer instante en el que el capital de la aseguradora se vuelve negativo.

La notación utilizada tiene el siguiente significado:

- $\tau$: tiempo de ruina.
- $t\geq 0$: se analiza el proceso desde el instante inicial.
- $U(t)<0$: condición que indica que el capital es negativo.
- $\inf$: ínfimo del conjunto de tiempos en los que se verifica la condición de ruina.

**Interpretación actuarial:**

La ruina ocurre cuando el capital disponible resulta insuficiente para afrontar las obligaciones de la aseguradora, de acuerdo con el modelo considerado.

Por ejemplo, si una compañía comienza con un capital de $1.000.000 y, luego de 3 años y 4 meses, experimenta un siniestro que provoca que sus reservas se vuelvan negativas por primera vez, entonces:

$$
\tau=3+\frac{4}{12}=3{,}333\text{ años}
$$

Si el capital nunca se vuelve negativo, se establece convencionalmente:

$$
\tau=\infty
$$

Es importante destacar que la ruina actuarial es un concepto matemático y no necesariamente implica la quiebra legal de la compañía.

---

## 3. Probabilidad de ruina en horizonte finito

La **probabilidad de ruina antes de un horizonte temporal $T$** se define como:

$$
\boxed{\psi(u,T)=\mathbb{P}(\tau\leq T\mid U(0)=u)}
$$

Esta expresión representa la probabilidad de que la aseguradora experimente una situación de ruina en algún instante anterior o igual al horizonte $T$, dado que comenzó con un capital inicial $u$.

Donde:

- $\psi(u,T)$: probabilidad de ruina en horizonte finito.
- $u$: capital inicial de la aseguradora.
- $T$: horizonte temporal de análisis.
- $\tau\leq T$: evento que indica que la ruina ocurre dentro del horizonte.
- $U(0)=u$: condición inicial del proceso.

### Ejemplo numérico

Supongamos que una compañía aseguradora comienza con un capital inicial de $10.000.000 y desea evaluar su riesgo de ruina durante los próximos 5 años.

Para ello, se simulan 10.000 trayectorias posibles del proceso de reservas mediante el método de Monte Carlo.

Los resultados hipotéticos son:

| Resultado | Simulaciones |
|---|---:|
| Trayectorias sin ruina | 9.200 |
| Trayectorias con ruina | 800 |
| Total | 10.000 |

La probabilidad estimada de ruina es:

$$
\hat{\psi}(10.000.000,5)
=\frac{800}{10.000}
=0{,}08
$$

Por lo tanto:

$$
\boxed{\hat{\psi}(10.000.000,5)=8\%}
$$

**Interpretación:**

Bajo las hipótesis del modelo utilizado, la aseguradora presenta una probabilidad estimada del 8% de experimentar una situación de ruina durante los próximos 5 años.

Esto no significa necesariamente que la compañía termine el quinto año con capital negativo.

Por ejemplo, podría haber experimentado una situación de ruina durante el segundo año y posteriormente haber recuperado un capital positivo.

Para el cálculo de la probabilidad de ruina, lo relevante es que el capital haya sido negativo **al menos una vez durante el horizonte considerado**.

---

## 4. Probabilidad de supervivencia

La **probabilidad de supervivencia** hasta el horizonte $T$ se define como:

$$
\boxed{\phi(u,T)=1-\psi(u,T)}
$$

Equivalentemente:

$$
\phi(u,T)=\mathbb{P}(\tau>T\mid U(0)=u)
$$

Representa la probabilidad de que el capital de la aseguradora no se vuelva negativo en ningún instante del horizonte analizado.

Retomando el ejemplo anterior:

$$
\phi(10.000.000,5)=1-0{,}08
$$

Por lo tanto:

$$
\boxed{\phi(10.000.000,5)=0{,}92=92\%}
$$

**Interpretación:**

La compañía aseguradora tiene una probabilidad estimada del 92% de no experimentar una situación de ruina durante los próximos 5 años.

La probabilidad de supervivencia constituye el complemento de la probabilidad de ruina:

$$
\boxed{\psi(u,T)+\phi(u,T)=1}
$$

---

## 5. Propiedades de la probabilidad de ruina

### 5.1. Relación con el capital inicial

En el modelo clásico, manteniendo constantes las demás condiciones, cuanto mayor sea el capital inicial de la aseguradora, menor o igual será su probabilidad de ruina.

Matemáticamente:

$$
u_1<u_2
\quad\Longrightarrow\quad
\psi(u_1,T)\geq\psi(u_2,T)
$$

**Interpretación:**

Una aseguradora con mayor capital inicial dispone de un mayor respaldo para absorber pérdidas ocasionadas por siniestros.

Por lo tanto, un incremento del capital inicial no aumenta su probabilidad de ruina, bajo las hipótesis del modelo.

### 5.2. Relación con el horizonte temporal

La probabilidad de ruina es una función no decreciente del horizonte temporal.

Es decir:

$$
T_1<T_2
\quad\Longrightarrow\quad
\psi(u,T_1)\leq\psi(u,T_2)
$$

**Interpretación:**

Cuanto mayor sea el período durante el cual se observa el proceso de reservas, mayores serán las oportunidades de que ocurra una situación de ruina.

Por ejemplo, la probabilidad de ruina durante los próximos 10 años no puede ser inferior a la probabilidad de ruina durante los próximos 5 años, si se considera el mismo proceso y capital inicial.

---

## 6. Diferencia entre ruina y capital negativo al final del horizonte

Es fundamental distinguir entre dos conceptos:

**a) Probabilidad de que el capital sea negativo al finalizar el horizonte:**

$$
\mathbb{P}(U(T)<0\mid U(0)=u)
$$

**b) Probabilidad de experimentar ruina en algún momento del horizonte:**

$$
\psi(u,T)=\mathbb{P}(\tau\leq T\mid U(0)=u)
$$

La primera expresión considera únicamente el capital en el instante final $T$.

La segunda considera toda la trayectoria del proceso de reservas desde el instante inicial hasta $T$.

Por lo tanto:

$$
\boxed{
\mathbb{P}(U(T)<0\mid U(0)=u)
\leq
\psi(u,T)
}
$$

Esta desigualdad se debe a que una aseguradora puede experimentar una situación de ruina durante el horizonte y posteriormente recuperar un capital positivo.

En cambio, si el capital es negativo en el instante final, necesariamente se habrá producido una situación de ruina dentro del horizonte.

---

## 7. Probabilidad de ruina en horizonte infinito

Además de estudiar la probabilidad de ruina durante un horizonte temporal finito, resulta relevante analizar qué ocurre cuando el horizonte de evaluación se extiende indefinidamente.

Se define la **probabilidad de ruina última** como:

$$
\boxed{
\psi(u)=\mathbb{P}(\tau<\infty\mid U(0)=u)
}
$$

Equivalentemente:

$$
\psi(u)=\lim_{T\to\infty}\psi(u,T)
$$

Esta magnitud representa la probabilidad de que la aseguradora experimente una situación de ruina en algún momento futuro, sin establecer un límite temporal.

En consecuencia:

$$
\boxed{\psi(u,T)\leq\psi(u)}
$$

La probabilidad de ruina en un horizonte finito nunca puede superar la probabilidad de ruina última.

---

## 8. Conclusión

La Teoría del Riesgo Actuarial permite evaluar la solvencia de una compañía aseguradora mediante el análisis de la evolución estocástica de su capital.

Los conceptos fundamentales son:

- **Proceso de reservas $U(t)$:** describe la evolución del capital de la aseguradora.
- **Tiempo de ruina $\tau$:** primer instante en el que el capital se vuelve negativo.
- **Probabilidad de ruina $\psi(u,T)$:** probabilidad de que ocurra ruina dentro de un horizonte temporal.
- **Probabilidad de supervivencia $\phi(u,T)$:** probabilidad de no experimentar ruina durante dicho horizonte.
- **Probabilidad de ruina última $\psi(u)$:** probabilidad de experimentar ruina en algún momento futuro.

La principal distinción consiste en que el análisis de ruina no se limita a evaluar el capital disponible al finalizar un período, sino que considera toda su trayectoria temporal.

Estos conceptos constituyen la base para estudiar la suficiencia de capital, la solvencia de las compañías aseguradoras y la aplicación de métodos estocásticos, como las simulaciones de Monte Carlo, a la gestión del riesgo actuarial.



## 9. Distribución condicional del tiempo de ruina

Recordemos que la **probabilidad de ruina última** se define como:

$$
\psi(u)=\mathbb{P}(\tau<\infty\mid U(0)=u)
$$

Esta expresión representa la probabilidad de que una compañía aseguradora, con capital inicial $u$, experimente una situación de ruina en algún momento futuro.

Cuando:

$$
\boxed{\psi(u)<1}
$$

existe una probabilidad positiva de que la aseguradora nunca experimente ruina.

Por lo tanto, la probabilidad de supervivencia indefinida es:

$$
\mathbb{P}(\tau=\infty\mid U(0)=u)=1-\psi(u)
$$

Por ejemplo, si:

$$
\psi(u)=0.20
$$

entonces existe una probabilidad del 20% de que la compañía eventualmente experimente ruina y una probabilidad del 80% de que nunca lo haga.

Sin embargo, además de conocer la probabilidad de ruina última, resulta interesante estudiar **cuándo ocurre la ruina entre aquellas trayectorias que eventualmente experimentan ese evento**.

Para ello, se define la distribución condicional del tiempo de ruina:

$$
\boxed{
\mathbb{P}(\tau\leq t\mid\tau<\infty,U(0)=u)
=
\frac{\psi(u,t)}{\psi(u)}
}
$$

Esta expresión representa la probabilidad de que la ruina ocurra antes o en el instante $t$, condicionada a que la aseguradora eventualmente experimente ruina. Se supone que $\psi(u)>0$.

### Interpretación mediante probabilidad condicional

Recordemos que, para dos eventos $A$ y $B$:

$$
\mathbb{P}(A\mid B)=
\frac{\mathbb{P}(A\cap B)}{\mathbb{P}(B)}
$$

En nuestro caso, definimos:

- $A=\{\tau\leq t\}$: la ruina ocurre antes de $t$ o en ese instante.
- $B=\{\tau<\infty\}$: la ruina ocurre en algún momento futuro.

Como toda trayectoria que experimenta ruina antes de $t$ necesariamente experimenta ruina en algún momento, se cumple:

$$
A\subseteq B
$$

Por lo tanto:

$$
\mathbb{P}(A\cap B)=\mathbb{P}(A)
$$

Aplicando la definición de probabilidad condicional, y manteniendo fijo el capital inicial $u$:

$$
\mathbb{P}(\tau\leq t\mid\tau<\infty,U(0)=u)
=
\frac{
\mathbb{P}(\tau\leq t\mid U(0)=u)
}{
\mathbb{P}(\tau<\infty\mid U(0)=u)
}
$$

Finalmente:

$$
\boxed{
F_{\tau\mid\tau<\infty}(t)
=
\frac{\psi(u,t)}{\psi(u)}
}
$$

Esta expresión constituye una **función de distribución acumulada del tiempo de ruina condicionada a que la ruina ocurra**.

## 10. Ejemplo numérico

Supongamos que una compañía aseguradora presenta las siguientes probabilidades de ruina, obtenidas mediante un modelo actuarial:

| Horizonte temporal | Probabilidad de ruina |
|---|---:|
| 1 año | 5% |
| 3 años | 12% |
| 5 años | 16% |
| Horizonte infinito | 20% |

La probabilidad de ruina última es:

$$
\psi(u)=0.20
$$

Y la probabilidad de ruina antes de los primeros 5 años es:

$$
\psi(u,5)=0.16
$$

La probabilidad condicional de experimentar ruina antes de 5 años, dado que eventualmente ocurrirá, es:

$$
\mathbb{P}(\tau\leq5\mid\tau<\infty)
=
\frac{\psi(u,5)}{\psi(u)}
$$

Reemplazando:

$$
\mathbb{P}(\tau\leq5\mid\tau<\infty)
=
\frac{0.16}{0.20}
=
0.80
$$

Por lo tanto:

$$
\boxed{
\mathbb{P}(\tau\leq5\mid\tau<\infty)=80\%
}
$$

**Interpretación actuarial:**

Entre las trayectorias que eventualmente experimentan ruina, el 80% presenta dicho evento durante los primeros 5 años.

Para comprenderlo, imaginemos que se analizan 1.000 trayectorias hipotéticas del proceso de reservas:

| Resultado | Cantidad de trayectorias |
|---|---:|
| Nunca experimentan ruina | 800 |
| Experimentan ruina durante los primeros 5 años | 160 |
| Experimentan ruina después de 5 años | 40 |
| **Total** | **1.000** |

De las 1.000 trayectorias, solamente 200 experimentan ruina en algún momento.

De esas 200 trayectorias ruinosas, 160 lo hacen antes de finalizar el quinto año.

Por lo tanto:

$$
\frac{160}{200}=0.80
$$

La diferencia fundamental es que:

- **Probabilidad incondicional de ruina antes de 5 años:** 16%, considerando todas las trayectorias.
- **Probabilidad condicional de ruina antes de 5 años:** 80%, considerando únicamente las trayectorias que eventualmente experimentan ruina.

## 11. Utilidad actuarial de la distribución condicional

La distribución condicional del tiempo de ruina permite diferenciar compañías aseguradoras que presentan una misma probabilidad de ruina última, pero cuya vulnerabilidad se manifiesta en horizontes temporales diferentes.

Consideremos dos compañías aseguradoras, A y B, que presentan la misma probabilidad de ruina última:

$$
\psi_A(u)=\psi_B(u)=0.20
$$

Sin embargo, sus probabilidades de ruina antes de los primeros 5 años son distintas:

$$
\psi_A(u,5)=0.18
$$

$$
\psi_B(u,5)=0.04
$$

Calculamos las probabilidades condicionales.

**Compañía A:**

$$
\mathbb{P}_A(\tau\leq5\mid\tau<\infty)
=
\frac{0.18}{0.20}
=
0.90
$$

**Compañía B:**

$$
\mathbb{P}_B(\tau\leq5\mid\tau<\infty)
=
\frac{0.04}{0.20}
=
0.20
$$

Resumiendo:

| Indicador | Compañía A | Compañía B |
|---|---:|---:|
| Probabilidad de ruina última | 20% | 20% |
| Probabilidad de ruina antes de 5 años | 18% | 4% |
| Probabilidad condicional de ruina antes de 5 años | 90% | 20% |

### Interpretación

**Compañía A:**

Entre las trayectorias que eventualmente experimentan ruina, el 90% presenta ese evento durante los primeros 5 años.

Esto significa que sus eventos de ruina se encuentran fuertemente concentrados en el corto plazo.

**Compañía B:**

Entre las trayectorias que eventualmente experimentan ruina, solamente el 20% presenta ese evento durante los primeros 5 años.

Por lo tanto, una proporción importante de sus eventos de ruina se concentra en horizontes más largos.

**Conclusión actuarial:**

Aunque ambas compañías tienen la misma probabilidad de ruina última, presentan perfiles temporales de riesgo muy diferentes.

La compañía A presenta una mayor vulnerabilidad en el corto plazo, mientras que la compañía B concentra una mayor proporción de sus eventos de ruina en el largo plazo.

Esta distinción es útil para:

- Evaluar la suficiencia de capital.
- Planificar estrategias de solvencia.
- Identificar vulnerabilidades de corto y largo plazo.
- Comparar carteras aseguradoras con diferente estructura de riesgos.
- Diseñar estrategias de mitigación y transferencia de riesgos.

## 12. Propiedades matemáticas de la distribución condicional

Definimos la función de distribución acumulada condicional del tiempo de ruina como:

$$
F_{\tau\mid\tau<\infty}(t)
=
\frac{\psi(u,t)}{\psi(u)}
$$

Siempre que $\psi(u)>0$, esta función presenta las siguientes propiedades.

### 12.1. Sus valores pertenecen al intervalo [0,1]

$$
0\leq F_{\tau\mid\tau<\infty}(t)\leq1
$$

Esto se debe a que:

$$
0\leq\psi(u,t)\leq\psi(u)
$$

### 12.2. Es una función no decreciente

Si:

$$
t_1<t_2
$$

entonces:

$$
F_{\tau\mid\tau<\infty}(t_1)
\leq
F_{\tau\mid\tau<\infty}(t_2)
$$

Esto ocurre porque, al aumentar el horizonte temporal, crece o se mantiene la proporción acumulada de trayectorias ruinosas que ya experimentaron el evento.

### 12.3. Converge a 1 en horizonte infinito

Se cumple que:

$$
\lim_{t\to\infty}
F_{\tau\mid\tau<\infty}(t)
=
1
$$

En efecto:

$$
\lim_{t\to\infty}
\frac{\psi(u,t)}{\psi(u)}
=
\frac{\psi(u)}{\psi(u)}
=
1
$$

Esto resulta intuitivo porque estamos considerando exclusivamente las trayectorias que eventualmente experimentan ruina.

## 13. Supervivencia condicional del tiempo de ruina

También puede resultar interesante calcular la probabilidad de que la ruina ocurra después de un horizonte determinado, sabiendo que eventualmente ocurrirá.

Esta probabilidad se expresa como:

$$
\mathbb{P}(\tau>t\mid\tau<\infty,U(0)=u)
$$

Utilizando el complemento de la función de distribución:

$$
\boxed{
\mathbb{P}(\tau>t\mid\tau<\infty,U(0)=u)
=
1-\frac{\psi(u,t)}{\psi(u)}
}
$$

Por ejemplo, si:

$$
\psi(u,5)=0.16
$$

y:

$$
\psi(u)=0.20
$$

entonces:

$$
\mathbb{P}(\tau>5\mid\tau<\infty)
=
1-\frac{0.16}{0.20}
$$

Por lo tanto:

$$
\boxed{
\mathbb{P}(\tau>5\mid\tau<\infty)=0.20
}
$$

**Interpretación:**

Entre las trayectorias que eventualmente experimentan ruina, el 20% lo hace después de los primeros 5 años.

Esta probabilidad no debe confundirse con la probabilidad de supervivencia incondicional:

$$
\phi(u,t)=1-\psi(u,t)
$$

La primera considera solamente las trayectorias que eventualmente experimentan ruina, mientras que la segunda considera todas las trayectorias posibles.

## 14. Tiempo esperado hasta la ruina

Una extensión natural del análisis consiste en estudiar el tiempo esperado hasta la ruina, condicionado a que esta ocurra.

Se define:

$$
\boxed{
\mathbb{E}[\tau\mid\tau<\infty,U(0)=u]
}
$$

Esta esperanza representa el tiempo promedio hasta la ruina, considerando únicamente aquellas trayectorias que eventualmente experimentan dicho evento.

Cuando la esperanza es finita, puede calcularse mediante:

$$
\mathbb{E}[\tau\mid\tau<\infty,U(0)=u]
=
\int_0^\infty
\mathbb{P}(\tau>t\mid\tau<\infty,U(0)=u)\,dt
$$

Reemplazando la expresión de supervivencia condicional:

$$
\boxed{
\mathbb{E}[\tau\mid\tau<\infty,U(0)=u]
=
\int_0^\infty
\left(
1-\frac{\psi(u,t)}{\psi(u)}
\right)dt
}
$$

**Interpretación actuarial:**

Mientras que la probabilidad de ruina última permite determinar la posibilidad de que la compañía eventualmente experimente ruina, el tiempo esperado hasta la ruina permite analizar cuánto tiempo transcurre, en promedio, hasta que dicho evento se produce entre las trayectorias ruinosas.

Esta medida complementa el análisis probabilístico y proporciona información relevante para la gestión temporal de los riesgos.

## 15. Conclusión

El análisis de ruina puede abordarse desde diferentes perspectivas:

**1. Probabilidad de ruina en horizonte finito:**

$$
\psi(u,T)=\mathbb{P}(\tau\leq T\mid U(0)=u)
$$

Permite determinar la probabilidad de que la compañía experimente ruina dentro de un período determinado.

**2. Probabilidad de ruina última:**

$$
\psi(u)=\mathbb{P}(\tau<\infty\mid U(0)=u)
$$

Permite determinar la probabilidad de que la ruina ocurra en algún momento futuro.

**3. Distribución condicional del tiempo de ruina:**

$$
F_{\tau\mid\tau<\infty}(t)
=
\frac{\psi(u,t)}{\psi(u)}
$$

Permite estudiar cuándo ocurre la ruina, considerando solamente aquellas trayectorias que eventualmente experimentan dicho evento.

**4. Tiempo esperado hasta la ruina:**

$$
\mathbb{E}[\tau\mid\tau<\infty,U(0)=u]
$$

Permite cuantificar el tiempo promedio hasta la ocurrencia de la ruina, condicionado a que esta efectivamente suceda.

En conjunto, estas herramientas permiten estudiar no solamente **la probabilidad de que una compañía aseguradora experimente ruina, sino también la distribución temporal de ese evento**.

Esto resulta especialmente relevante para la evaluación de solvencia, la administración del capital y la gestión integral de riesgos actuariales.
